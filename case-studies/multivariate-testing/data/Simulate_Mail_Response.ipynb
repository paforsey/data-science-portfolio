## Data Disclosure

- **This entire dataset is synthetic.** Every record below is generated from a random number generator, not collected from any real mail campaign
- Ground-truth response rates are set deliberately equal for Creative A and Creative B (1.00% each), so that the "no significant difference" result in the main notebook reflects a true null by construction, not an artifact of chance
- This notebook exists only to keep the raw data separate from the analysis notebook, consistent with how the other portfolio notebooks in `selected-work/` keep their source data in an adjacent `data/` folder
- Output: `mail_response_part1.csv`, one row per simulated mail recipient, imported directly by `Direct_Mail_Multivariate_Testing.ipynb`


### Initialize Environment

```python
# Core Libraries
import numpy as np
import pandas as pd

# Reproducibility
rng = np.random.default_rng(7)
```

## Simulate Individual-Level Mail Response

```python
# Simulate Mail Response: Creative A vs Creative B (Individual-Level Records)
N_PER_CELL = 500_000

# Ground truth: genuinely equal response rates - a true null by construction
true_rate = {'A': 0.0100, 'B': 0.0100}

records = []
for creative, rate in true_rate.items():
    n_resp = rng.binomial(N_PER_CELL, rate)
    responded = np.zeros(N_PER_CELL, dtype = int)
    responded[:n_resp] = 1
    rng.shuffle(responded)
    records.append(pd.DataFrame({'creative': creative, 'responded': responded}))

mail_response_df = pd.concat(records, ignore_index = True)

print('Total records:', len(mail_response_df))
print(mail_response_df.groupby('creative')['responded'].agg(['count', 'sum', 'mean']))
```

## Save to CSV

```python
# Save Synthetic Data
mail_response_df.to_csv('mail_response_part1.csv', index = False)
print('Saved mail_response_part1.csv:', mail_response_df.shape)
```
